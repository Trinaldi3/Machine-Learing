```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid")


df = pd.read_csv('tips.csv')

plt.figure(figsize=(8, 5))
sns.regplot(
    data=df,
    x='total_bill',
    y='tip',
    scatter_kws={'alpha': 0.6, 'color': '#2b5c8f'},
    line_kws={'color': '#e74c3c', 'linewidth': 2},
)
plt.title('1. Hubungan Total Bill dan Tip', fontsize=12, fontweight='bold')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()
```

```python
import os
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid")

if os.path.exists('sample_data/tips.csv'):
  df = pd.read_csv('sample_data/tips.csv')
elif os.path.exists('tips.csv'):
  df = pd.read_csv('tips.csv')
else:
  df = sns.load_dataset('tips')

plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df, x='day', y='total_bill', hue='day', palette='Set2', legend=False
)
plt.title(
    '2. Distribusi Total Bill Berdasarkan Hari', fontsize=12, fontweight='bold'
)
plt.xlabel('Hari')
plt.ylabel('Total Bill ($)')
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_csv('tips.csv')

plt.figure(figsize=(8, 5))
sns.barplot(
    data=df, x='time', y='tip', hue='sex', palette='coolwarm', errorbar=None
)
plt.title(
    '3. Rata-rata Tip Berdasarkan Waktu Makan & Gender',
    fontsize=12,
    fontweight='bold',
)
plt.xlabel('Waktu Makan')
plt.ylabel('Rata-rata Tip ($)')
plt.legend(title='Gender')
plt.show()
```

```python
import os
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid")


if os.path.exists('sample_data/tips.csv'):
  df = pd.read_csv('sample_data/tips.csv')
elif os.path.exists('tips.csv'):
  df = pd.read_csv('tips.csv')
else:
  df = sns.load_dataset('tips')

df['tip_pct'] = (df['tip'] / df['total_bill']) * 100

plt.figure(figsize=(8, 5))
sns.violinplot(
    data=df,
    x='smoker',
    y='tip_pct',
    hue='smoker',
    palette='Pastel1',
    inner='quartile',
    legend=False,
)
plt.title(
    '4. Variasi Persentase Tip Berdasarkan Status Merokok',
    fontsize=12,
    fontweight='bold',
)
plt.xlabel('Status Merokok')
plt.ylabel('Persentase Tip (%)')
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_csv('tips.csv')

fig, ax1 = plt.subplots(figsize=(8, 5))

sns.barplot(
    data=df,
    x='size',
    y='total_bill',
    color='#3498db',
    ax=ax1,
    alpha=0.7,
    errorbar=None,
)
ax1.set_title(
    '5. Pengaruh Ukuran Rombongan terhadap Total Bill & Tip',
    fontsize=12,
    fontweight='bold',
)
ax1.set_xlabel('Ukuran Rombongan')
ax1.set_ylabel('Rata-rata Total Bill ($)', color='#3498db')

ax2 = ax1.twinx()
sns.lineplot(
    data=df,
    x=df['size'] - 1,
    y='tip',
    color='#e67e22',
    marker='o',
    linewidth=2.5,
    ax=ax2,
    errorbar=None,
)
ax2.set_ylabel('Rata-rata Tip ($)', color='#e67e22')
ax2.grid(False)

plt.show()
```

```python
import pandas as pd

# reading the database
data = pd.read_csv("tips.csv")

# printing the top 10 rows
display(data.head(10))
```

```python
# import matplotlib to be able to use scatter plot and other kind of plots
import matplotlib.pyplot as plt

# Scatter plot with day (x) against tip (y)
plt.scatter(data['day'], data['tip'])

# Adding Title to the Plot
plt.title("Scatter Plot")

# Setting the X and Y labels
plt.xlabel('Day')
plt.ylabel('Tip')

# show the plot
plt.show()
```

```python
#import matplotlib to be able to use scatter plot and other kind of plots
import matplotlib.pyplot as plt

# Scatter plot with day (x) against tip (y)
plt.scatter(data['day'], data['tip'], c=data['size'],
            s=data['total_bill'])

# Adding Title to the Plot
plt.title("Scatter Plot")

# Setting the X and Y labels
plt.xlabel('Day')
plt.ylabel('Tip')

#use colorbar function
plt.colorbar()

#show the plot
plt.show()
```
